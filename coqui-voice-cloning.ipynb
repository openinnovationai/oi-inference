### Coqui voice cloning via OICM inference

```python
import requests
import os
```

```python
api_key = os.getenv('OI_API_KEY') # Your valid api key
deployment_id = os.getenv('OI_DEPLOYMENT_ID') # deployment id
# example: https://inference.{oicm_host}/models/{deployment_id}/proxy/vllm/v1
# oicm_host: is the platform url ex: develop.openinnovation.ai

base_url = f"https://inference.prod.openinnovation.ai/models/{deployment_id}/proxy/v1"

# base_url = "http://localhost:8080/v1"

headers = {
    "Authorization": f"Bearer {api_key}"
}
```

```python
with open("./cloning-input.mp3", "rb") as f:
    audio_data = f.read()

    files = {
        "speaker_voice_file": ("cloning-input.mp3", audio_data, "audio/mpeg")
    }

    data = {
        "prompt": "*laughing*",
        "speaker_language": "en",
    }

    # response = requests.post(
    #     f"{base_url}/generate-speech",
    #     headers=headers,
    #     json=data,
    # )

    response = requests.post(
        f"{base_url}/generate-speech",
        headers=headers,
        data=data,
        files=files
    )
```

```python
response_data = response.json()

audio = response_data["audio"]
sampling_rate = response_data["sampling_rate"]
```

```python
import IPython.display as ipd
import base64

# render the audio in the notebook
ipd.Audio(base64.b64decode(audio), rate=sampling_rate)

# # write the audio to a file
# with open("cloned-voice.wav", "wb") as f:
#     f.write(base64.b64decode(audio))
```
